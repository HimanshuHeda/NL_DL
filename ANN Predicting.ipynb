## Predicting the Next CGPA Value Using ANN
This example uses the CGPA column from your dataset to train an ANN model that predicts the next CGPA value in the sequence.

```python
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Load dataset
df = pd.read_csv('./Dataset/Admission_Predict.csv')

# Prepare data: Use CGPA to predict next CGPA
X = df['CGPA'][:-1].values.reshape(-1, 1)  # All except last
y = df['CGPA'][1:].values  # All except first

# Build ANN model
model = Sequential()
model.add(Dense(8, activation='relu', input_dim=1))
model.add(Dense(1))  # Output layer

model.compile(optimizer='adam', loss='mean_squared_error')
model.fit(X, y, epochs=100, verbose=1)

# Predict next CGPA for last value
last_cgpa = df['CGPA'].iloc[-1]
next_cgpa = model.predict(np.array([[last_cgpa]]))
print(f"Predicted next CGPA after {last_cgpa}: {next_cgpa[0][0]:.2f}")
```

## Predicting the Next GRE Score Using ANN
This example uses the GRE Score column from your dataset to train an ANN model that predicts the next GRE Score in the sequence.

```python
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Load dataset
df = pd.read_csv('./Dataset/Admission_Predict.csv')

# Prepare data: Use GRE Score to predict next GRE Score
X_gre = df['GRE Score'][:-1].values.reshape(-1, 1)  # All except last
y_gre = df['GRE Score'][1:].values  # All except first

# Build ANN model
model_gre = Sequential()
model_gre.add(Dense(8, activation='relu', input_dim=1))
model_gre.add(Dense(1))  # Output layer

model_gre.compile(optimizer='adam', loss='mean_squared_error')
model_gre.fit(X_gre, y_gre, epochs=100, verbose=1)

# Predict next GRE Score for last value
last_gre = df['GRE Score'].iloc[-1]
next_gre = model_gre.predict(np.array([[last_gre]]))
print(f"Predicted next GRE Score after {last_gre}: {next_gre[0][0]:.2f}")
```
